### This notebook includes all possible pipe diagrams to compare for the teleportation

Pipe diagram A

```python
from tqec import BlockGraph
from tqec.utils.position import Position3D

g = BlockGraph("Teleportation")

cubes = [
    (Position3D(0, 0, 0), "ZXZ", "In"),   # vertical pipe going up
    (Position3D(0, 0, 1), "ZXX", ""),      # junction: vertical from below, horizontal to the side
    (Position3D(0, 1, 1), "ZXX", ""),      # junction: horizontal from above, vertical going up  
    (Position3D(0, 1, 2), "ZXZ", "Out"),   # vertical pipe coming in
]

for pos, kind, label in cubes:
    g.add_cube(pos, kind, label)

pipes = [(0, 1), (1, 2), (2, 3)]
for p0, p1 in pipes:
    g.add_pipe(cubes[p0][0], cubes[p1][0])

g.view_as_html()
```

```python
from tqec.computation.prism import BasisPrism, Position3DHex, ZXPrism
from tqec.computation.pipe_prism import PrismPipeKind
from tqec.computation.prism_graph import PrismGraph

g = PrismGraph("two-patches")

prisms = [
    (Position3DHex(0,0,-1), "ZN", "init"),
    (Position3DHex(0,0,0), "NX", "left"),
    (Position3DHex(1,1,0), "XN", "right"),
    (Position3DHex(1,1,1), "NZ", "final"),
]

for pos, kind, label in prisms:
    g.add_prism(pos, kind, label)

pipe_kind = PrismPipeKind(hor = BasisPrism.N, ver = BasisPrism.N)
g.add_pipe(prisms[0][0], prisms[1][0], pipe_kind)

pipe_kind = PrismPipeKind(hor = BasisPrism.X, ver = BasisPrism.Z)
g.add_pipe(prisms[1][0], prisms[2][0], pipe_kind)

pipe_kind = PrismPipeKind(hor = BasisPrism.N, ver = BasisPrism.N)
g.add_pipe(prisms[2][0], prisms[3][0], pipe_kind)

g.view_as_html()
```

### simulation

```python
from tqec import BlockGraph
from tqec.utils.position import Position3D
import sinter
import matplotlib.pyplot as plt

g_sc = BlockGraph("Teleportation")
cubes = [
    # Initial logical qubit
    (Position3D(0, 0, 0), "ZXZ", "Init_Z"),

    # Intermediate teleportation structure
    (Position3D(0, 0, 1), "ZXX", ""),
    (Position3D(0, 1, 1), "ZXX", ""),

    # Output logical qubit
    (Position3D(0, 1, 2), "ZXZ", "Out"),
]


for pos, kind, label in cubes:
    g_sc.add_cube(pos, kind, label)

pipes = [
    (0, 1),
    (1, 2),
    (2, 3),
]

for p0, p1 in pipes:
    g_sc.add_pipe(cubes[p0][0], cubes[p1][0])

correlation_surfaces = g_sc.find_correlation_surfaces()
observables = [correlation_surfaces[0]]
```

```python
import numpy as np
from pathlib import Path

from tqec import NoiseModel
from tqec.simulation.simulation import start_simulation_using_sinter

path="/mnt/c/Users/Laura/Documents/Codes_and_Repos/tqec_cococo/tqec/temp-laura/teleportation_superdense"

d_lst = [3,5]
ks = [(d - 1) / 2 for d in d_lst]
p_values = list(np.logspace(-4, -1, 10))

max_shots = 10_000
max_errors = 100

if __name__ == "__main__":
    stats_sc = start_simulation_using_sinter(
        g_sc,
        ks=ks,  # k values for the code distance
        ps=p_values,  # error rates
        noise_model_factory=NoiseModel.uniform_depolarizing,  # noise model
        manhattan_radius=2,  # parameter for automatic detector computation
        observables=[correlation_surfaces[0]],  # observable of interest
        decoders=["pymatching"],
        num_workers=8,
        max_shots=max_shots,
        max_errors=max_errors,
        save_resume_filepath=Path(path+"sc_a_resume.csv"),
        database_path=Path(path+f"sc_a_maxshots_{max_shots}_maxerrors_{max_errors}.pkl"),
    )
```

```python
#path_cc = "/mnt/c/Users/Laura/Documents/Codes_and_Repos/tqec_cococo/tqec/temp-laura/teleportation_superdense/tele_sinter_a_260524_114846_maxshots10000000_maxerrors500.csv"
path_cc = "/mnt/c/Users/Laura/Documents/Codes_and_Repos/tqec_cococo/tqec/temp-laura/teleportation_superdense/tele_sinter_a_260524_090530.csv"
stats_cc = sinter.read_stats_from_csv_files(path_cc)
```

```python
from itertools import chain
from tqec.simulation.plotting.inset import plot_observable_as_inset

stats_sc_flat = list(chain.from_iterable(stats_sc))
stats_cc_flat = stats_cc  # already flat from read_stats_from_csv_files

zx_graph = g_sc.to_zx_graph()

fig, ax = plt.subplots()

sinter.plot_error_rate(
    ax=ax,
    stats=stats_sc_flat,  # pass directly
    x_func=lambda stat: stat.json_metadata["p"],
    group_func=lambda stat: f"Surface Code d={round(stat.json_metadata['d'])}",
    plot_args_func=lambda index, group, stats: {"linestyle": ":"},
)

sinter.plot_error_rate(
    ax=ax,
    stats=stats_cc_flat,  # pass directly
    x_func=lambda stat: stat.json_metadata["p"],
    group_func=lambda stat: f"Color Code d={round(stat.json_metadata['d'])}",
    plot_args_func=lambda index, group, stats: {"linestyle": "-"},
)

plot_observable_as_inset(ax, zx_graph, correlation_surfaces[0])
ax.grid(axis="both")
ax.legend()
ax.loglog()
ax.set_xlabel("Physical Error Rate")
ax.set_ylabel("Logical Error Rate")
plt.show()
```

```python
import numpy as np
import matplotlib.colors as mcolors
import matplotlib.pyplot as plt
from itertools import chain
from tqec.simulation.plotting.inset import plot_observable_as_inset

# Get sorted unique d values across both datasets
all_d_values = sorted(set(
    round(stat.json_metadata["d"]) for stat in stats_sc_flat + stats_cc_flat
))

# Build rainbow colormap: one color per d
cmap = plt.cm.cool
d_to_color = {
    d: cmap(i / (len(all_d_values) - 1))
    for i, d in enumerate(all_d_values)
}

def lighten(color, factor=0.5):
    """Blend color with white by factor (0=original, 1=white)."""
    r, g, b, a = mcolors.to_rgba(color)
    return (1 - (1 - r) * (1 - factor),
            1 - (1 - g) * (1 - factor),
            1 - (1 - b) * (1 - factor), a)

fig, ax = plt.subplots()

sinter.plot_error_rate(
    ax=ax,
    stats=stats_sc_flat,
    x_func=lambda stat: stat.json_metadata["p"],
    group_func=lambda stat: f"Surface Code d={round(stat.json_metadata['d'])}",
    plot_args_func=lambda index, group, stats: {
        "linestyle": ":",
        "color": lighten(d_to_color[round(stats[0].json_metadata["d"])], factor=0.5),
    },
)

sinter.plot_error_rate(
    ax=ax,
    stats=stats_cc_flat,
    x_func=lambda stat: stat.json_metadata["p"],
    group_func=lambda stat: f"Color Code d={round(stat.json_metadata['d'])}",
    plot_args_func=lambda index, group, stats: {
        "linestyle": "-",
        "color": d_to_color[round(stats[0].json_metadata["d"])],
    },
)

plot_observable_as_inset(ax, zx_graph, correlation_surfaces[0])
ax.grid(axis="both")
ax.legend()
ax.loglog()
ax.set_xlabel("Physical Error Rate")
ax.set_ylabel("Logical Error Rate")
plt.show()
```
